## Ejercicio 1. Probabilidad condicional

Se lanza un dado equilibrado y se desea calcular la probabilidad de obtener un 2, sabiendo que salió un número par.

**Datos:**

- Espacio muestral: {1, 2, 3, 4, 5, 6}.
- Números pares: {2, 4, 6}.
- Casos favorables: {2}.

**Fórmula de probabilidad condicional:**

P(A | B) = P(A ∩ B) / P(B)

Como sabemos que el resultado es par, solo consideramos tres posibilidades: 2, 4 y 6. De estas tres posibilidades, únicamente una corresponde al número 2.

Por lo tanto, la probabilidad es 1/3, equivalente al 33.33%.

```python
# Ejercicio 1: Probabilidad condicional

# Resultados posibles al lanzar un dado
dado = [1, 2, 3, 4, 5, 6]

# Seleccionamos los números pares
numeros_pares = [numero for numero in dado if numero % 2 == 0]

# Buscamos los resultados que cumplen ambas condiciones
favorables = [numero for numero in numeros_pares if numero == 2]

# Calculamos la probabilidad condicional
probabilidad = len(favorables) / len(numeros_pares)

# Mostramos los resultados
print("Números pares:", numeros_pares)
print("Casos favorables:", favorables)
print("Probabilidad de obtener un 2 dado que salió par:",
      round(probabilidad, 4))
print("Porcentaje:", round(probabilidad * 100, 2), "%")
```

Conclusión: La probabilidad condicional permite calcular la posibilidad de un evento cuando ya conocemos información previa. En este ejercicio, al saber que salió un número par, la probabilidad de obtener un 2 es de 33.33%.

## Ejercicio 2. Probabilidad condicional con tres dados

Se lanzan tres dados equilibrados y queremos calcular la probabilidad de que la suma de sus resultados sea 9, sabiendo que las tres caras son distintas.

**Datos:**

- Cada dado tiene 6 caras.
- Se consideran únicamente los resultados donde los tres dados muestran números diferentes.
- El evento favorable ocurre cuando los tres números son distintos y su suma es 9.

**Fórmula:**

P(A | B) = P(A ∩ B) / P(B)

Donde:

- A: la suma de los tres dados es 9.
- B: los tres dados muestran caras distintas.
- A ∩ B: los tres resultados son diferentes y suman 9.

Para resolverlo, contamos todos los resultados posibles que cumplen la condición de tener tres caras distintas y después contamos cuántos de ellos suman 9.

```python
# Ejercicio 2: Probabilidad condicional con tres dados

from itertools import product

# Generamos todas las combinaciones posibles de tres dados
resultados = list(product(range(1, 7), repeat=3))

# Seleccionamos los casos donde las tres caras son distintas
caras_distintas = [
    resultado for resultado in resultados
    if len(set(resultado)) == 3
]

# De los anteriores, seleccionamos los que suman 9
suma_nueve = [
    resultado for resultado in caras_distintas
    if sum(resultado) == 9
]

# Calculamos la probabilidad condicional
probabilidad = len(suma_nueve) / len(caras_distintas)

# Mostramos los resultados
print("Total de resultados posibles:", len(resultados))
print("Resultados con tres caras distintas:", len(caras_distintas))
print("Resultados con caras distintas que suman 9:", len(suma_nueve))
print("Probabilidad:", round(probabilidad, 4))
print("Porcentaje:", round(probabilidad * 100, 2), "%")
```

**Conclusión:**

La probabilidad de que la suma de los tres dados sea 9, sabiendo que las tres caras son distintas, es del **15%**.

En Python se generaron las 216 combinaciones posibles, se seleccionaron las 120 combinaciones con resultados diferentes y se encontraron 18 que suman 9. Por lo tanto, la probabilidad es 18/120 = 0.15, equivalente al 15%.

## Ejercicio 3. Probabilidad de lluvia dado que está nublado

De un total de 100 días, 40 son nublados y de esos días, en 30 llueve. Se desea calcular la probabilidad de que llueva sabiendo que el día está nublado.

**Datos:**

- Total de días: 100
- Días nublados: 40
- Días nublados en los que llueve: 30

**Fórmula de probabilidad condicional:**

P(L | N) = P(L ∩ N) / P(N)

Donde:

- L: el día presenta lluvia.
- N: el día está nublado.

Como solo tomamos en cuenta los 40 días nublados, dividimos los 30 días en los que llovió entre los 40 días nublados.

```python
# Ejercicio 3: Probabilidad de lluvia dado que está nublado

# Datos del problema
total_dias = 100
dias_nublados = 40
dias_nublados_con_lluvia = 30

# Calculamos la probabilidad condicional
probabilidad = dias_nublados_con_lluvia / dias_nublados

# Mostramos los resultados
print("Total de días:", total_dias)
print("Días nublados:", dias_nublados)
print("Días nublados con lluvia:", dias_nublados_con_lluvia)
print("Probabilidad de lluvia dado que está nublado:",
      round(probabilidad, 2))
print("Porcentaje:", round(probabilidad * 100, 2), "%")
```

**Conclusión:**

La probabilidad de que llueva dado que el día está nublado es del **75%**. Esto significa que, de cada 100 días observados, cuando el día está nublado, en el 75% de esos casos se presenta lluvia.

El cálculo se realiza dividiendo los 30 días nublados con lluvia entre los 40 días nublados: 30 / 40 = 0.75.
